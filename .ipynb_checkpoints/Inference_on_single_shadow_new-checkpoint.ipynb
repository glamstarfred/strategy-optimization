```python
def simulate_classical_shadow(all_probs, measurement_df, qubit_num, shadow_id, t_index):
    """
    Generate a single classical shadow snapshot at a specified time index
    using predicted measurement outcome probabilities.

    This function simulates one round of local Pauli measurements for an
    n-qubit system. For each qubit, it:

        1. Retrieves the Pauli observable (X, Y, or Z) assigned to the given
           shadow_id from `measurement_df`.
        2. Reads the predicted probability p = P(+1) from `all_probs`.
        3. Samples a measurement outcome from a Bernoulli distribution:
              +1 with probability p,
              -1 with probability 1 - p.
        4. Stores the (pauli_label, outcome) pair.

    Parameters
    ----------
    all_probs : torch.Tensor or np.ndarray
        Array of predicted probabilities with shape:
            (qubit_num, num_shadows, num_time_points)
        where:
            all_probs[q, s, t] = P(outcome = +1 | qubit=q,
                                                shadow=s,
                                                time=t)

    measurement_df : pandas.DataFrame
        DataFrame containing the Pauli measurement settings.
        Required columns:
            - "qubit"     : int
            - "shadow_id" : int
            - "pauli"     : str ('X', 'Y', or 'Z')
        Each (qubit, shadow_id) pair must appear exactly once.

    qubit_num : int
        Total number of qubits in the system.

    shadow_id : int
        Index specifying which classical shadow configuration to use.

    t_index : int
        Time index selecting which probability slice to sample from.

    Returns
    -------
    shadow : list of [str, int]
        A list of length `qubit_num`, where each element is:
            [pauli_label, outcome]
        with outcome from {+1, -1}.
        This represents one classical shadow measurement snapshot.

    """
    shadow = []

    for q in range(qubit_num):
        # get the Pauli observable for this qubit/shadow
        obs_row = measurement_df[
            (measurement_df["qubit"] == q) &
            (measurement_df["shadow_id"] == shadow_id)
        ]
        if obs_row.empty:
            raise ValueError(f"No observable found for qubit {q}, shadow {shadow_id}")

        pauli_label = obs_row['pauli'].values[0]  # e.g., 'X', 'Y', 'Z'

        # get predicted probability for +1 outcome at t_index
        p = all_probs[q, shadow_id, t_index].item()
        # print(p)

        # simulate outcome: +1 with prob p, -1 with prob 1-p
        outcome = 1 if random.random() < p else -1

        shadow.append([pauli_label, outcome])

    return shadow
```

```python
############################# Definition of useful constants ##############################
############################################################################################
# definition for all the basis and pauli matrix
paulis = {
    "X": qt.sigmax(),
    "Y": qt.sigmay(),
    "Z": qt.sigmaz()
}

ket0 = qt.basis(2, 0)
ket1 = qt.basis(2, 1)

pauli_eigen = {
    'Z': {+1: ket0, -1: ket1},
    'X': {+1: (ket0 + ket1).unit(),
          -1: (ket0 - ket1).unit()},
    'Y': {+1: (ket0 + 1j*ket1).unit(),
          -1: (ket0 - 1j*ket1).unit()}
}

############################################################################################
shadow_size = 100
qubit_num  = 2
```

```python
# Define a quantum state plus
all_qubit = [(ket0+ket1).unit() for i in range(qubit_num)]
psi = qt.tensor(all_qubit)

# difine operator as \sigma_x \otimes \sigma_x \cdots for simplicity. estimations should be around 1
operator = qt.tensor(qt.sigmax(), qt.qeye(2))

rho = psi * psi.dag()
tlist = np.linspace(0, 2*np.pi, 2000)
####################################################################################
# definition of quantum dynamic
H = qt.tensor(qt.sigmaz(), qt.sigmaz())

# damping Hamiltonian model
gamma = 0.1
c_ops = [np.sqrt(gamma) * qt.tensor(qt.sigmaz(), qt.qeye(2))]

# result = qt.mesolve(H, psi, tlist, c_ops) # with damping
result = qt.mesolve(H, psi, tlist, []) # without damping

# expect = qt.expect(operator, result.states)
```

# Generation of data
This generate random outcome of pauli measurements at sparse time index. Datas are saved in measurement_df for further use.

```python
indices = np.linspace(0, len(tlist) - 1, 20, dtype=int)
picked_times = tlist[indices]

# create variable name list for classical shadow
obs_list = []
for i in range(shadow_size):
    obs = create_random_pauli_obs(qubit_num)
    obs_list.append(obs)
records = []

for idx in indices:
    rho = result.states[idx] * result.states[idx].dag()
    time = tlist[idx]
    for shadow_id in range(shadow_size):
        obs = obs_list[shadow_id]
        measurements = pauli_measurement(rho, obs, qubit_num)

        for qubit, (pauli_label, outcome) in enumerate(measurements):
            records.append({
                # "time_idx": time_idx,
                "time": time,
                "shadow_id": shadow_id,
                "qubit": qubit,
                "pauli": pauli_label,
                "outcome": outcome,
            })

measurement_df = pd.DataFrame(records)

print(measurement_df)
```

# Bayesian inference on shadow 
This part does Bayesian inference on separated shadow data. 

```python
# preallocate storage
all_probs = torch.zeros(qubit_num, shadow_size, len(tlist))
all_means = torch.zeros(qubit_num, shadow_size, len(tlist))
all_vars = torch.zeros(qubit_num, shadow_size, len(tlist))


# convert tlist once
test_x = torch.tensor(tlist, dtype=torch.float32).unsqueeze(-1)  # (T,1)

for q in range(qubit_num):
    # select all data for this qubit
    qubit_data = measurement_df[measurement_df["qubit"] == q]

    # get unique shadows for this qubit
    shadows = qubit_data["shadow_id"].unique()

    # process all shadows in one loop over shadows (still vectorized internally)
    for s in shadows:
        
        # filter data for this shadow
        shadow_data = qubit_data[qubit_data["shadow_id"] == s]

        # convert to torch tensors
        train_x = torch.tensor(
            shadow_data["time"].to_numpy(),
            dtype=torch.float32
        ).unsqueeze(-1)
    
        train_y = torch.tensor(
            shadow_data["outcome"].to_numpy(),
            dtype=torch.float32
        )

        # train model and get predictions
        # ---- Train GP classifier ----
        model, likelihood = train_gp_classifier(train_x, train_y)
    
        # ---- Predict ----
        prob_star, mean_star, var_star = predict_gp_classifier(
            model, likelihood, test_x
        )

        # store in preallocated tensor
        all_probs[q, s, :] = prob_star
        # all_means[q, s, :] = mean_star
        # all_vars[q, s, :] = var_star

# the means and variance are not useful currently
```

```python
# simulate classical shadow on test time indexs
test_indices = np.linspace(0, len(tlist) - 1, 100, dtype=int)
test_times = tlist[test_indices]

# preallocate
num_times = len(test_indices)
shot_at_test_times = [[None for _ in range(shadow_size)] for _ in range(num_times)]

for t_idx, idx in enumerate(test_indices):
    for s_id in range(shadow_size):
        # simulate shadow and convert to Qobj
        shot_data = simulate_classical_shadow(all_probs, measurement_df, qubit_num=2, shadow_id=s_id, t_index=idx)
        # store directly in preallocated list
        shot_at_test_times[t_idx][s_id] = shot_data
```

```python
# classical shadow estimation
estimation_list = []
for i in range(len(test_indices)):
    shot_data = shot_at_test_times[i]
    shadow = create_classical_shadow(shot_data)
    value = estimation_mean(shadow, operator)
    estimation_list.append(value)
```

# Accuracy Check

```python
# ------------------------------------------------------------
# 2. Single-shot shadow reconstruction
# ------------------------------------------------------------
def single_shadow_rho_df(measurement_rows, qubit_num):
    """
    measurement_rows: dataframe slice corresponding to one shadow_id at one time
    """
    local_estimators = []

    for q in range(qubit_num):
        row = measurement_rows[measurement_rows["qubit"] == q].iloc[0]
        pauli = row["pauli"]
        outcome = row["outcome"]

        ket = pauli_eigen[pauli][outcome]
        proj = ket * ket.dag()

        local_estimators.append(3 * proj - qt.qeye(2))

    return qt.tensor(local_estimators)


# ------------------------------------------------------------
# 3. Classical shadow expectation estimation
# ------------------------------------------------------------
def classical_shadow_expectation_df(measurement_df, observable, qubit_num):
    """
    Computes classical shadow estimation of observable.

    Returns
    -------
    pd.DataFrame with columns:
        time
        expectation
    """

    results = []

    grouped = measurement_df.groupby(["time", "shadow_id"])

    temp_storage = {}

    for (time, shadow_id), group in grouped:
        rho_hat = single_shadow_rho_df(group, qubit_num)
        value = (observable * rho_hat).tr().real

        if time not in temp_storage:
            temp_storage[time] = []

        temp_storage[time].append(value)

    # Average over shadows for each time
    for time, values in temp_storage.items():
        results.append({
            "time": time,
            "expectation": np.mean(values)
        })

    # Convert to DataFrame and sort
    result_df = pd.DataFrame(results)
    result_df = result_df.sort_values("time").reset_index(drop=True)

    return result_df
```

```python
shadow_estimates = classical_shadow_expectation_df(measurement_df, operator, qubit_num)

print(shadow_estimates)
```

```python
from botorch.models import SingleTaskGP
from botorch.fit import fit_gpytorch_mll
from gpytorch.mlls import ExactMarginalLogLikelihood
from gpytorch.kernels import MaternKernel, ScaleKernel


def bayesian_regression_matern(
    result_df,
    nu=2.5,
    lengthscale_prior=None,
    train_noise=True,
    num_test_points=300,
    device="cpu",
):
    """
    Perform Bayesian regression using BoTorch with Matérn kernel.

    Parameters
    ----------
    result_df : pd.DataFrame
        Must contain columns:
            'time'
            'expectation'

    nu : float
        Smoothness parameter of Matérn kernel (0.5, 1.5, 2.5)

    train_noise : bool
        Whether to learn noise level

    num_test_points : int
        Number of prediction points

    Returns
    -------
    dict containing:
        model
        train_x
        train_y
        test_x
        posterior_mean
        posterior_var
    """

    # ----------------------------
    # Prepare training data
    # ----------------------------
    train_x = torch.tensor(
        result_df["time"].values,
        dtype=torch.double,
        device=device
    ).unsqueeze(-1)

    train_y = torch.tensor(
        result_df["expectation"].values,
        dtype=torch.double,
        device=device
    ).unsqueeze(-1)

    # ----------------------------
    # Define Matérn Kernel
    # ----------------------------
    covar_module = ScaleKernel(
        MaternKernel(nu=nu)
    )

    model = SingleTaskGP(
        train_X=train_x,
        train_Y=train_y,
        covar_module=covar_module
    )

    # ----------------------------
    # Fit model
    # ----------------------------
    mll = ExactMarginalLogLikelihood(model.likelihood, model)
    fit_gpytorch_mll(mll)

    model.eval()

    # ----------------------------
    # Prediction grid
    # ----------------------------
    t_min = train_x.min().item()
    t_max = train_x.max().item()

    test_x = torch.linspace(
        t_min,
        t_max,
        num_test_points,
        dtype=torch.double,
        device=device
    ).unsqueeze(-1)

    with torch.no_grad():
        posterior = model.posterior(test_x)
        mean = posterior.mean.squeeze(-1)
        var = posterior.variance.squeeze(-1)

    return {
        "model": model,
        "train_x": train_x,
        "train_y": train_y,
        "test_x": test_x,
        "posterior_mean": mean,
        "posterior_var": var,
    }
```

```python
gp_results = bayesian_regression_matern(shadow_estimates)

mean = gp_results["posterior_mean"].cpu().numpy()
var = gp_results["posterior_var"].cpu().numpy()
test_x = gp_results["test_x"].cpu().numpy()
```

```python
def compute_trust_zone(all_probs, pauli, measurement_df, qubit_idx, confidence=0.95):
    """
    Compute trust zone over shadows restricted to a specific Pauli.

    Parameters
    ----------
    all_probs : (Q, S, T)
    pauli : str  ("X", "Y", or "Z")
    measurement_df : pandas DataFrame
        Must contain columns: 'pauli' and 'shadow_index'
    confidence : float

    Returns
    -------
    mean_overall : (Q, T)
    lower_bound  : (Q, T)
    upper_bound  : (Q, T)
    var_shadow   : (Q, T)
    """
    tensor_shadow_ids = sorted(measurement_df["shadow_id"].unique())

    shadow_id_to_tensor_index = {
        sid: idx for idx, sid in enumerate(tensor_shadow_ids)
    }

    shadow_ids = measurement_df.loc[
        (measurement_df["pauli"] == pauli) &
        (measurement_df["qubit"] == qubit_idx),
        "shadow_id"
    ].unique()
    
    tensor_indices = torch.tensor(
        [shadow_id_to_tensor_index[sid] for sid in shadow_ids],
        dtype=torch.long
    )
    
    probs_selected = all_probs[:, tensor_indices, :]
    
    # shadow_indices = measurement_df.loc[
    #     measurement_df["pauli"] == pauli, "shadow_id"
    # ].unique()
    # print(len(shadow_indices))

    # shadow_indices = torch.tensor(shadow_indices, dtype=torch.long)

    # probs_selected = all_probs[:, shadow_indices, :]   # (Q, S_p, T)

    # probs_selected = probs_selected[qubit_index]

    mean_overall = probs_selected.mean(dim=1)           # (Q, T)
    var_shadow   = probs_selected.var(dim=1, unbiased=False)

    std_total = torch.sqrt(var_shadow)

    if confidence == 0.95:
        z = 1.96
    elif confidence == 0.68:
        z = 1.0
    else:
        from scipy.stats import norm
        z = norm.ppf((1 + confidence) / 2)

    lower_bound = mean_overall - z * std_total
    upper_bound = mean_overall + z * std_total

    return mean_overall, lower_bound, upper_bound, var_shadow, probs_selected
```
