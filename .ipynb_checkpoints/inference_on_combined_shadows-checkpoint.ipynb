```python
def simulate_classical_shadow_on_combined_shadow(all_probs, measurement_df, qubit_num, shadow_id, t_index):
    """
    Generate a classical shadow snapshot at a given time index using
    Pauli-resolved probability predictions (combined-shadow format).

    This function simulates one round of local Pauli measurements for an
    n-qubit system. Unlike the standard shadow format where probabilities
    are indexed by (qubit, shadow_id, time), here probabilities are stored
    in a *Pauli-resolved* structure:

        all_probs[q, p, t] = P(outcome = +1 | qubit=q,
                                              Pauli=p,
                                              time=t)

    where:
        p = 0 → X
        p = 1 → Y
        p = 2 → Z

    For each qubit, the function:
        1. Retrieves the Pauli observable assigned by `measurement_df`
           for the specified shadow_id.
        2. Maps the Pauli label to its corresponding index p ∈ {0,1,2}.
        3. Extracts the predicted probability P(+1) from `all_probs`.
        4. Samples a measurement outcome:
               +1 with probability p,
               -1 with probability 1 - p.
        5. Appends the (pauli_label, outcome) pair to the shadow list.

    Parameters
    ----------
    all_probs : torch.Tensor or np.ndarray
        Probability tensor of shape:
            (qubit_num, 3, num_time_points)
        storing predicted probabilities for +1 outcomes of
        X, Y, Z measurements per qubit and time index.

    measurement_df : pandas.DataFrame
        DataFrame defining the Pauli measurement settings.
        Required columns:
            - "qubit"     : int
            - "shadow_id" : int
            - "pauli"     : str ('X', 'Y', or 'Z')
        Each (qubit, shadow_id) pair must appear exactly once.

    qubit_num : int
        Total number of qubits.

    shadow_id : int
        Identifier selecting which measurement configuration to simulate.

    t_index : int
        Time index selecting the probability slice from `all_probs`.

    Returns
    -------
    shadow : list of [str, int]
        List of length `qubit_num`, where each element is:
            [pauli_label, outcome]
        with outcome ∈ {+1, -1}.
        Represents one classical shadow measurement snapshot.
    """
    
    shadow = []

    for q in range(qubit_num):
        # get the Pauli observable for this qubit/shadow
        obs_row = measurement_df[
            (measurement_df["qubit"] == q) &
            (measurement_df["shadow_id"] == shadow_id)
        ]
        if obs_row.empty:
            raise ValueError(f"No observable found for qubit {q}, shadow {shadow_id}")

        pauli_label = obs_row['pauli'].values[0]  # e.g., 'X', 'Y', 'Z'
        if pauli_label == 'X':
            p_number = 0
        elif pauli_label == 'Y':
            p_number = 1
        elif pauli_label == 'Z':
            p_number = 2

        # get predicted probability for +1 outcome at t_index
        p = all_probs[q, p_number, t_index].item()
        # print(p)

        # simulate outcome: +1 with prob p, -1 with prob 1-p
        outcome = 1 if random.random() < p else -1

        shadow.append([pauli_label, outcome])

    return shadow
```

```python
############################################################################################
# definition for all the basis and pauli matrix
paulis = {
    "X": qt.sigmax(),
    "Y": qt.sigmay(),
    "Z": qt.sigmaz()
}

ket0 = qt.basis(2, 0)
ket1 = qt.basis(2, 1)

pauli_eigen = {
    'Z': {+1: ket0, -1: ket1},
    'X': {+1: (ket0 + ket1).unit(),
          -1: (ket0 - ket1).unit()},
    'Y': {+1: (ket0 + 1j*ket1).unit(),
          -1: (ket0 - 1j*ket1).unit()}
}

############################################################################################
# definition of qubits and state, here the state is n tensor product of |+> 
qubit_num  = 2
shadow_size = 100
```

```python
# Define a quantum state |+>
all_qubit = [(ket0+ket1).unit() for i in range(qubit_num)]
psi = qt.tensor(all_qubit)

# difine operator as \sigma_x \otimes \sigma_x \cdots for simplicity. estimations should be around 1
operator = qt.tensor(qt.sigmax(), qt.qeye(2))

rho = psi * psi.dag()
tlist = np.linspace(0, 2*np.pi, 2000)
####################################################################################
# definition of quantum dynamic
H = qt.tensor(qt.sigmaz(), qt.sigmaz())

gamma = 0.1
c_ops = [np.sqrt(gamma) * qt.tensor(qt.sigmaz(), qt.qeye(2))]
# result = qt.mesolve(H, psi, tlist, c_ops) # with damping
result = qt.mesolve(H, psi, tlist, []) # without damping
```

# Generation of Data
This part is the same as the part in inference_on_seperate_shadow

```python
indices = np.linspace(0, len(tlist) - 1, 20, dtype=int)
picked_times = tlist[indices]

shadow_size = 100
qubit_num  = 2

obs_list = []
for i in range(shadow_size):
    obs = create_random_pauli_obs(qubit_num)
    obs_list.append(obs)
records = []

for idx in indices:
    rho = result.states[idx] * result.states[idx].dag()
    time = tlist[idx]
    for shadow_id in range(shadow_size):
        obs = obs_list[shadow_id]
        measurements = pauli_measurement(rho, obs, qubit_num)

        for qubit, (pauli_label, outcome) in enumerate(measurements):
            records.append({
                # "time_idx": time_idx,
                "time": time,
                "shadow_id": shadow_id,
                "qubit": qubit,
                "pauli": pauli_label,
                "outcome": outcome,
            })

measurement_df = pd.DataFrame(records)

print(measurement_df)
```

# Bayesian Inference on shadow
Unlike inference_on_seperate_shadow.py, this time Bayesian inference is applied to all the shadow with the same qubit number and observable name (X,Y,Z). So there's no loop on shadow_id

```python
variables = ['X','Y','Z']
all_probs = torch.zeros(qubit_num, len(variables), len(tlist))

# convert tlist once
test_x = torch.tensor(tlist, dtype=torch.float32).unsqueeze(-1)
# test_x_tensor = torch.tensor(tlist, dtype=torch.float32).unsqueeze(-1)  # (T,1)

for q in range(qubit_num):
    # select all data for this qubit
    qubit_data = measurement_df[measurement_df["qubit"] == q]

    # process all shadows in one loop over shadows (still vectorized internally)
    for s, pauli in enumerate(variables):
        # print(pauli)
        
        # filter data for this shadow
        shadow_data = qubit_data[qubit_data["pauli"] == pauli]

        # convert to torch tensors
        train_x = torch.tensor(
            shadow_data["time"].to_numpy(),
            dtype=torch.float32
        ).unsqueeze(-1)
    
        train_y = torch.tensor(
            shadow_data["outcome"].to_numpy(),
            dtype=torch.float32
        )

        # train model and get predictions
        # ---- Train GP classifier ----
        model, likelihood = train_gp_classifier(train_x, train_y)
    
        # ---- Predict ----
        prob_star, mean_star, var_star = predict_gp_classifier(
            model, likelihood, test_x
        )

        # store in preallocated tensor
        all_probs[q, s, :] = prob_star

```

```python
# simulate classical shadow on test time indexs
test_indices = np.linspace(0, len(tlist) - 1, 200, dtype=int)
test_times = tlist[test_indices]

# preallocate
num_times = len(test_indices)
shot_at_test_times = [[None for _ in range(shadow_size)] for _ in range(num_times)]

for t_idx, idx in enumerate(test_indices):
    for s_id in range(shadow_size):
        # simulate shadow and convert to Qobj
        shot_data = simulate_classical_shadow_on_combined_shadow(all_probs, measurement_df, qubit_num=2, shadow_id=s_id, t_index=idx)
        # store directly in preallocated list
        shot_at_test_times[t_idx][s_id] = shot_data
```

```python
# classical shadow estimation
estimation_list = []
for i in range(len(test_indices)):
    shot_data = shot_at_test_times[i]
    shadow = create_classical_shadow(shot_data)
    value = estimation_mean(shadow, operator)
    estimation_list.append(value)
```

```python
fig, axes = plt.subplots(1, 1)
expect = qt.expect(operator, result.states)
axes.plot(test_times, estimation_list, label='Classical Shadow Prediction')
axes.plot(tlist, expect, label=r"$\left<\sigma_x I\right>$")
axes.set_xlabel(r"$t$", fontsize=20)
axes.legend(loc=2)
plt.show()
plt.close()
```

# Other Plots

```python
# this part filters the shadow on the first qubit with measurement on X, and apply Bayesian inference on all of them
data = measurement_df[
    (measurement_df["qubit"] == 0) &
    (measurement_df["pauli"] == 'X')
]

print("Total data points:", len(data))

train_x = torch.tensor(
    data["time"].to_numpy(),
    dtype=torch.float32
).unsqueeze(-1)

train_y = torch.tensor(
    data["outcome"].to_numpy(),
    dtype=torch.float32
)

test_x = torch.tensor(tlist, dtype=torch.float32).unsqueeze(-1)

model, likelihood = train_gp_classifier(train_x, train_y)

prob_star, mean_star, var_star = predict_gp_classifier(
    model, likelihood, test_x
)
```

```python
mean_np  = mean_star.detach().cpu().numpy()
prob_np = prob_star.detach().cpu().numpy()
var_np = var_star.detach().cpu().numpy()


# mu, var are latent GP outputs
sigma_latent = np.sqrt(var_np)

# 95% CI in latent space
lower_f = mean_np - 1.96 * sigma_latent
upper_f = mean_np + 1.96 * sigma_latent

# transform to probability space
lower_prob = norm.cdf(lower_f)
upper_prob = norm.cdf(upper_f)


plus = (ket0 + ket1).unit()
# plus = (ket0 + 1j*ket1).unit()
# plus = ket0
proj = qt.tensor(plus * plus.dag(), qt.qeye(2))
true_prob = [qt.expect(proj, psi) for psi in result.states]

# ---- plot ----
plt.figure(figsize=(8, 5))

plt.plot(tlist, prob_np, label='Bayesian Prediction')

plt.fill_between(
    tlist,
    lower_prob,
    upper_prob,
    alpha=0.3,
    label='95% credible interval'
)

plt.plot(tlist, true_prob, label='True')
plt.scatter(
    picked_times,
    np.zeros_like(picked_times),
    color = 'black',
    marker='x',
    s=80,
    label='Chosen time points'
)

plt.xlabel("t")
plt.ylabel("Predicted probability P(X=1 | t)")
plt.title("Bayesian GP Probability Prediction on P(X=1)")
plt.ylim(-0.1, 1.1)
plt.legend(loc='upper left')
plt.show()
plt.close()
```
